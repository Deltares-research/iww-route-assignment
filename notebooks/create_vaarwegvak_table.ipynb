```python
import pandas as pd
import geopandas as gpd

from pathlib import Path
```

```python
src_dir = Path('~').expanduser() / 'data/ivs/iww-route-assignment'
```

```python
df = pd.read_csv(src_dir / 'Telpunten_20260717.csv', sep=';')
```

```python
df['OBJECT ID'].unique().shape
```

```python
unique_telpunt_df = df.sort_values(by=['OBJECT ID', 'isrsCode']).groupby('OBJECT ID').first()
unique_telpunt_gdf = gpd.GeoDataFrame(unique_telpunt_df, geometry=gpd.points_from_xy(unique_telpunt_df['Longitude'], unique_telpunt_df['Latitude']))
unique_telpunt_gdf = unique_telpunt_gdf.set_crs('EPSG:4326')
unique_telpunt_gdf = unique_telpunt_gdf.to_crs('EPSG:28992')
unique_telpunt_gdf = unique_telpunt_gdf.rename_geometry("point_geometry")
```

```python
edges_gdf = gpd.read_file('/Users/hemert/projects/iww-route-assignment-sito-is/fis-crawler/output/euris-enriched/edges.geojson')
edges_gdf = edges_gdf.to_crs('EPSG:28992')
```

```python
edges_gdf
```

```python
joined_gdf = unique_telpunt_gdf.sjoin_nearest(edges_gdf, how='left', distance_col='distance_join').reset_index()
joined_gdf.sort_values(by=['OBJECT ID', 'distance_join']).groupby('OBJECT ID').first().sort_values(by='distance_join')
joined_gdf["geometry"] = joined_gdf["index_right"].map(edges_gdf.geometry)
joined_gdf = joined_gdf.set_geometry("geometry")
joined_gdf = gpd.GeoDataFrame(
    joined_gdf,
    geometry="geometry",
    crs=edges_gdf.crs,
)
joined_gdf["point_wkt"] = joined_gdf["point_geometry"].to_wkt()

joined_gdf = joined_gdf.drop(columns=["point_geometry"])
joined_gdf.to_csv(src_dir / 'euris_with_telpunt.csv')
joined_gdf.to_file(src_dir / 'euris_with_telpunt.gpkg')
```
