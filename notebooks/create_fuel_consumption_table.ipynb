```python
def get_classes_between(cemt_class_1, cemt_class_2):
    cemt_index_1 = cemt_order.index(cemt_class_1)
    cemt_index_2 = cemt_order.index(cemt_class_2)

    first_cemt_index = min(cemt_index_1, cemt_index_2) + 1
    last_cemt_index = max(cemt_index_1, cemt_index_2)

    cemt_classes = cemt_order[first_cemt_index:last_cemt_index]
    return cemt_classes
```

```python
df = pd.DataFrame(vessel_data)
df['vessel_cemt_class'] = df['vessel_rws_class'].map(rws_to_cemt)
df
```

```python
value_columns = [
    "diesel_light",
    "diesel_medium",
    "diesel_heavy",
]

new_rows = []

for rws_class, group in df.groupby("vessel_rws_class"):
    group = group.copy()

    # Exclude non-CEMT classes such as "VIc"
    group = group[group["waterway_cemt_class"].isin(cemt_order)]

    group["_cemt_index"] = group["waterway_cemt_class"].map(
        {cemt: i for i, cemt in enumerate(cemt_order)}
    )
    group = group.sort_values("_cemt_index")

    # Iterate over consecutive available rows
    for (_, lower_row), (_, upper_row) in zip(
        group.iloc[:-1].iterrows(),
        group.iloc[1:].iterrows(),
    ):
        lower_class = lower_row["waterway_cemt_class"]
        upper_class = upper_row["waterway_cemt_class"]

        missing_classes = get_classes_between(lower_class, upper_class)

        total_steps = (
            cemt_order.index(upper_class)
            - cemt_order.index(lower_class)
        )

        for missing_class in missing_classes:
            step = (
                cemt_order.index(missing_class)
                - cemt_order.index(lower_class)
            )
            fraction = step / total_steps

            new_row = lower_row.drop("_cemt_index").to_dict()
            new_row["waterway_cemt_class"] = missing_class

            for column in value_columns:
                new_row[column] = (
                    lower_row[column]
                    + fraction * (upper_row[column] - lower_row[column])
                )

            new_rows.append(new_row)

filled_df = pd.concat(
    [
        df,
        pd.DataFrame(new_rows),
    ],
    ignore_index=True,
)

filled_df["_cemt_index"] = filled_df["waterway_cemt_class"].map(
    {cemt: i for i, cemt in enumerate(cemt_order)}
)

filled_df = (
    filled_df
    .sort_values(["vessel_rws_class", "_cemt_index"], na_position="last")
    .drop(columns="_cemt_index")
    .reset_index(drop=True)
)
```
